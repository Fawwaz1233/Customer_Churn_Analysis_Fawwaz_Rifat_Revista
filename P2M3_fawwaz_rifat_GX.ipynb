```python
# Mmebuat data konteks

from great_expectations.data_context import FileDataContext

context = FileDataContext.create(project_root_dir='./')
```

```python
# Memeberikan nama Datasource
datasource_name = 'csv-customer-churn-2'
datasource = context.sources.add_pandas(datasource_name)

# Memberi nama data asset
asset_name = 'customer-churn'
path_to_data = 'P2M3_fawwaz_rifat_data_clean.csv'
asset = datasource.add_csv_asset(asset_name, filepath_or_buffer=path_to_data)

# Membuat batch request
batch_request = asset.build_batch_request()
```

```python
# Mmebuat expectation suite
expectation_suite_name = 'expectation-churn-dataset'
context.add_or_update_expectation_suite(expectation_suite_name)

# Membuat validator menggunakan expectation suite
validator = context.get_validator(
    batch_request = batch_request,
    expectation_suite_name = expectation_suite_name
)

# Mengecek validator
validator.head()
```

## **To be unique**

```python
# Expectation 1 : Kolom 'customerid' harus unik

validator.expect_column_values_to_be_unique('customerid')
```

## **To be between min_value and max_value**

```python
# Expectation 2 : Kolom 'tenure' tidak boleh kurang dari 0 atau lebih dari 12 bulan

validator.expect_column_values_to_be_between(
    column='tenure', min_value=0, max_value=12
)
```

## **To be in set**

```python
# Expectation 3 : Kolom 'exited' harus ada dalam data asset

validator.expect_column_to_exist(column='exited')
```

## **To be in type list**

```python
# Expectation 4 : Kolom 'balance' harus dalam bentuk interger atau float

validator.expect_column_values_to_be_in_type_list('balance', ['integer', 'float'])
```

## **To equal set**

```python
# Expectation 5 : Kolom 'card_type' hanya boleh terdiri dari 4 kategori
validator.expect_column_distinct_values_to_equal_set(
    "card_type",
    ['DIAMOND', 'GOLD', 'SILVER', 'PLATINUM']
)
```

## **To be of type**

```python
# Expectation 6 : Kolom 'balance' harus dalam bentuk float
validator.expect_column_values_to_be_of_type("estimatedsalary", "float")
```

## **To equal column count**

```python
# Expectation 7 : Jumlah kolom dalam data asset harus berjumlah 18
validator.expect_table_column_count_to_equal(18)
```
